  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>

```python
#202432051_NailahAmoritaSyaqraRafiun
import os
import re
from collections import Counter
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432051_NailahAmoritaSyaqraRafiun
# 1. Muat dataset
if os.path.exists('IMDB Dataset.csv'):
    df = pd.read_csv('IMDB Dataset.csv')
else:
    import kagglehub
    path = kagglehub.dataset_download('lakshmi25npathi/imdb-dataset-of-50k-movie-reviews')
    df = pd.read_csv(os.path.join(path, 'IMDB Dataset.csv'))
print('Ukuran data (baris, kolom):', df.shape)
print(df['sentiment'].value_counts())
df['label'] = (df['sentiment'] == 'positive').astype(int)
# 2. Pembersihan data & pra-pemrosesan NLP
def clean_text(text):
    text = text.lower()
    text = re.sub(r'<.*?>', ' ', text)
    text = re.sub(r'[^a-z\s]', ' ', text)
    return text.split()
df['tokens'] = df['review'].apply(clean_text)
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42, stratify=df['label'])
counter = Counter(w for toks in train_df['tokens'] for w in toks)
MAX_VOCAB = 10000
vocab = {'<pad>': 0, '<unk>': 1}
for w, _ in counter.most_common(MAX_VOCAB - 2):
    vocab[w] = len(vocab)
MAX_LEN = 200
def encode(tokens):
    ids = [vocab.get(w, 1) for w in tokens][:MAX_LEN]
    return [0] * (MAX_LEN - len(ids)) + ids
# 3. Konversi ke PyTorch Tensor
X_train = torch.tensor([encode(t) for t in train_df['tokens']], dtype=torch.long)
X_test = torch.tensor([encode(t) for t in test_df['tokens']], dtype=torch.long)
y_train = torch.tensor(train_df['label'].values, dtype=torch.float32)
y_test = torch.tensor(test_df['label'].values, dtype=torch.float32)
train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=128, shuffle=True)
test_loader = DataLoader(TensorDataset(X_test, y_test), batch_size=128)
print(X_train.shape, X_test.shape)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432051_NailahAmoritaSyaqraRafiun
class SequenceModel(nn.Module):
    def __init__(self, vocab_size, embed_dim=128, hidden_size=128, dropout=0.3):
        super(SequenceModel, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=0)
        self.lstm = nn.LSTM(embed_dim, hidden_size, batch_first=True)
        self.dropout = nn.Dropout(dropout)
        self.fc = nn.Linear(hidden_size, 1)
    def forward(self, x):
        emb = self.embedding(x)
        out, (h_n, c_n) = self.lstm(emb)
        last = self.dropout(h_n[-1])
        return self.fc(last).squeeze(1)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432051_NailahAmoritaSyaqraRafiun
# 1. Inisiasi hyperparameter
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
epochs = 8
learning_rate = 0.001
# 2. Inisiasi model, loss, optimizer
model = SequenceModel(len(vocab)).to(device)
criterion = nn.BCEWithLogitsLoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)
# 3. Riwayat training
train_losses = []
train_accs = []
test_accs = []
# 4. Training loop
for epoch in range(epochs):
    model.train()
    total_loss, correct = 0.0, 0
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        logits = model(xb)
        loss = criterion(logits, yb)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        total_loss += loss.item() * xb.size(0)
        correct += ((logits > 0).float() == yb).sum().item()
    train_losses.append(total_loss / len(train_loader.dataset))
    train_accs.append(correct / len(train_loader.dataset))
    model.eval()
    test_correct = 0
    with torch.no_grad():
        for xb, yb in test_loader:
            xb, yb = xb.to(device), yb.to(device)
            test_correct += ((model(xb) > 0).float() == yb).sum().item()
    test_accs.append(test_correct / len(test_loader.dataset))
    print(f'Epoch {epoch+1}/{epochs} | Loss: {train_losses[-1]:.4f} | Train Acc: {train_accs[-1]:.4f} | Test Acc: {test_accs[-1]:.4f}')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432051_NailahAmoritaSyaqraRafiun
# 1. Grafik penurunan loss training
plt.figure(figsize=(10, 4))
plt.plot(train_losses, color='#114F6A', linewidth=2, label='Training Loss')
plt.title('Grafik Penurunan Loss Selama Pelatihan')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.savefig('loss_curve.png', dpi=150, bbox_inches='tight')
plt.show()
# 2. Grafik akurasi latih vs uji
plt.figure(figsize=(10, 4))
plt.plot(train_accs, color='#114F6A', linewidth=2, label='Train Accuracy')
plt.plot(test_accs, color='#1FA4B1', linewidth=2, label='Test Accuracy')
plt.title('Riwayat Akurasi Latih vs Uji')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.savefig('acc_curve.png', dpi=150, bbox_inches='tight')
plt.show()
# 3. Prediksi pada data uji + skor sentimen
model.eval()
all_probs = []
with torch.no_grad():
    for xb, yb in test_loader:
        all_probs.append(torch.sigmoid(model(xb.to(device))).cpu())
all_probs = torch.cat(all_probs).numpy()
y_true = y_test.numpy().astype(int)
y_pred = (all_probs > 0.5).astype(int)
print(classification_report(y_true, y_pred, target_names=['negative', 'positive'], digits=4))
print('Confusion Matrix:')
print(confusion_matrix(y_true, y_pred))
# 4. Distribusi skor prediksi
plt.figure(figsize=(12, 5))
plt.hist(all_probs[y_true == 0], bins=30, alpha=0.7, color='#114F6A', label='Aktual Negative')
plt.hist(all_probs[y_true == 1], bins=30, alpha=0.7, color='#1FA4B1', label='Aktual Positive')
plt.axvline(0.5, color='red', linestyle='--', label='Threshold 0.5')
plt.title('Distribusi Skor Prediksi Sentimen pada Data Uji')
plt.xlabel('Skor Probabilitas Positif')
plt.ylabel('Jumlah Ulasan')
plt.legend()
plt.savefig('dist_pred.png', dpi=150, bbox_inches='tight')
plt.show()
# 5. Contoh prediksi per ulasan
for i in range(5):
    label = 'positive' if y_pred[i] == 1 else 'negative'
    actual = 'positive' if y_true[i] == 1 else 'negative'
    print(f'Skor={all_probs[i]:.4f} | Prediksi={label} | Aktual={actual} | {test_df["review"].iloc[i][:80]}...')
```
